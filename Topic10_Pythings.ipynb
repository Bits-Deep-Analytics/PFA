```python
# Packing: grouping values into a tuple
coordinates = (10, 20, 30)

# Unpacking: assigning elements to individual variables
x, y, z = coordinates

print(f"x: {x}, y: {y}, z: {z}")
# Output: x: 10, y: 20, z: 30
```

```python
# Unpacking in a for loop with a dictionary's .items()
pokemon = {'Rattata': 19, 'Machop': 66}
for name, number in pokemon.items(): # name and number are the loop variables
    print(f"{name}'s number is {number}")
# Output: Rattata's number is 19
#         Machop's number is 66
```

```python
data = (1, 2, 3, 4, 5)
a, b, *rest = data
# a is 1, b is 2, rest is [3, 4, 5]
```

```python
a = 1
b = 2
print(a,b)
a, b = b, a # a is now 2, b is now 1
print(a,b)
```

```python
fruits = ['apple', 'banana', 'cherry']
for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")
```
